```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import csv
```

```python
longitude,latitude,ave_tem=[],[],[]
```

```python
with open(r'C:\Users\12532\Documents\Python\My-climate-analysis\data\daily_weather_data.csv','r',encoding='utf-8') as f:
    reader=csv.DictReader(f)
    for row in reader:
        long_str=row['Longitude'].strip()
        lat_str=row['Latitude'].strip()
        tavg_str=row['tavg'].strip()
        if long_str != '' and lat_str != '' and tavg_str != '':
            long_f=float(long_str)
            longitude.append(long_f)
            lat_f=float(lat_str)
            latitude.append(lat_f)
            tavg_f=float(tavg_str)
            ave_tem.append(tavg_f)
```

```python
#累了画图明天再说
```

```python
#3-D plot
plt.figure(figsize=(10,6))
scatter=plt.scatter(longitude,latitude,c=ave_tem,cmap='coolwarm',s=50)
plt.colorbar(scatter,label='Temperature ℃')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.xticks(np.arange(-200, 200, 25), [f"{x}°E" for x in np.arange(-200, 200, 25)])
plt.yticks(np.arange(-80, 80, 10), [f"{y}°N" for y in np.arange(-80, 80, 10)])

plt.title('Temperature Distribution')
plt.show()
```
