```python
import csv
import math
```

```python
temperatures = [] #prepare a blank list
```

```python
with open(r'C:\Users\12532\Documents\Python\My-climate-analysis\data\daily_weather_data.csv','r',encoding='utf-8') as f:
    reader=csv.DictReader(f)
    for row in reader:
        temp_str=row['tavg'].strip()
        if temp_str != '':
            temp=float(temp_str)
            temperatures.append(temp)
Average_tem = sum(temperatures)/len(temperatures)
Max_tem=max(temperatures)
Min_tem=min(temperatures)

print(f'Average temperature in this area is {Average_tem:.1f}. The Maximum and minimum values are {Max_tem} and {Min_tem} respectively')
```
