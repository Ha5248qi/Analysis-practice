```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\12532\Documents\Python\My-climate-analysis\data\air-quality-india.csv",encoding='utf-8')
```

```python
df['Timestamp']=pd.to_datetime(df['Timestamp'])#pd.to_datetime(arg,errors='coerce')
#errors='coerce'跳过错误数据
```

```python
df.set_index('Timestamp',inplace=True)
```

```python
print(df.head(10))
```

```python
monthly_data = df['PM2.5'].resample('M').mean()
#Core code，链式调用，对象.操作1.操作2
```

```python
print(monthly_data.head)
```

```python
plt.figure(figsize=(10,6))
plt.plot(monthly_data, marker='o', linestyle='-', color='c')
plt.xlabel('Date')
plt.ylabel(f'Concentration of PM2.5 (μg/m³)')
plt.title('Change of the concentration of PM2.5 of India in 2017-2022')
#plt.axhline(y=数值, color='颜色', linestyle='线型', label='图例标签')
plt.axhline(y=12, color='g', linestyle='-')
plt.axhline(y=35, color='y', linestyle='-')
plt.axhline(y=55, color='pink', linestyle='-')
## 数显：plt.text(x=30, y=threshold + 1, s=f'PM2.5={threshold}', color='orange')
#ax.text(x=0.01, y=12 + 1, s='PM2.5=12 (Good)', color='green', transform=ax.get_yaxis_transform()获取 X 轴变换)
ax = plt.gca()
#Get Current Axes
plt.text(x=0.01, y=12 + 1, s=f'PM2.5={12,'Good'}', color='b',transform=ax.get_yaxis_transform())
plt.text(x=0.01, y=35 + 1, s=f'PM2.5={35,'Moderate'}', color='b',transform=ax.get_yaxis_transform())
plt.text(x=0.01, y=55 + 1, s=f'PM2.5={55,'Unhelthy'}', color='b',transform=ax.get_yaxis_transform())
plt.grid(True)
```
